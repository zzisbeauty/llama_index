```python
from llama_index.core import SummaryIndex
from llama_index.readers.notion import NotionPageReader
from IPython.display import Markdown, display
from dotenv import load_dotenv
import os

# 配置文件获取
load_dotenv('/app/docs/.env')
integration_token = os.getenv('notion')

# 设置环境变量
# os.environ["NOTION_INTEGRATION_TOKEN"] = "ntn_fan153489606154ERwjQ9EUiXTJQRmdfBzIjfEI3oLGaiwc6l"
# integration_token = os.getenv("NOTION_INTEGRATION_TOKEN")
```

```python
reader = NotionPageReader()  
  
# 尝试搜索页面  ID
try: 
    # 会匹配到很多有一些关键字的 page id
    page_ids = reader.search("开发LLM强化学习框架的心得")  
    print("找到的页面ID:", page_ids)  
except Exception as e:  
    print("错误:", e)
```

```python
import requests
  
# 搜索页面  
page_ids = reader.search("大开发LLM强化学习框架的心得规模")  
  
# 为每个页面ID获取标题  
for page_id in page_ids:  
    url = f"https://api.notion.com/v1/pages/{page_id}"  
    response = requests.get(url, headers=reader.headers)  
    data = response.json()  

    # 简单提取标题  
    title = "无标题"  
    if "properties" in data:  
        for prop_data in data["properties"].values():  
            if prop_data.get("type") == "title" and prop_data.get("title"):  
                title = prop_data["title"][0].get("plain_text", "无标题")  
                break

    print(f"ID: {page_id}")
    print(f"标题: {title}") 
    print("---")
```

```python
page_ids = ["9207d126-8842-45ce-bb00-b1d0d1d93e80"]
documents = NotionPageReader(integration_token=integration_token).load_data(
    page_ids=page_ids
)
```

```python
# 创建一个摘要索引（Summary Index）
# 1. 文档处理：将输入的 documents 转换为节点（nodes）
# 2. 存储结构：创建一个简单的列表数据结构，按顺序存储所有文档块
# 3. 索引构建：使用 IndexList 作为底层数据结构来组织这些节点
# todo: 可以基于这个索引进行查询，程序会根据 query 从索引中返回相关的 nodes
index = SummaryIndex.from_documents(documents)
```

You can also pass the id of a database to index all the pages in that database:

```python
database_ids = ["<database-id>"]

# https://developers.notion.com/docs/working-with-databases for how to find your database id

documents = NotionPageReader(integration_token=integration_token).load_data(
    database_ids=database_ids
)

print(documents)
```

```python
# set Logging to DEBUG for more detailed outputs
index = SummaryIndex.from_documents(documents)
query_engine = index.as_query_engine()
response = query_engine.query("<query_text>")
display(Markdown(f"<b>{response}</b>"))
```
